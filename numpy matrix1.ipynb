```python
import numpy as np

A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("Matrix A:\n", A)
print("Matrix B:\n", B)
```

```python
print("A + B =\n", A + B)
print("A - B =\n", A - B)
```

```python
  print("A dot B =\n", A.dot(B))
  print("A @ B=\n",A @ B)
```

```python
print("Element-wise multiply:\n",A*B)
```

```python
print("Transpose of A:\n", A.T)
```

```python
I = np.eye(3)
print("Identity Matrix:\n", I)
```

```python
Z = np.zeros((3,3))
O = np.ones((2,4))

print("Zero Matrix:\n", Z)
print("Ones Matrix:\n", O)
```

```python
R = np.random.randint(1, 10, (3,3))
print("Random Matrix:\n", R)
```

```python
A = np.array([[4, 7],
              [2, 6]])

inv_A = np.linalg.inv(A)
print("Inverse of A:\n", inv_A)
```

```python
x = np.arange(12)
M = x.reshape(3,4)
print("Reshaped Matrix:\n", M)
```

```python
A = np.ones((2,2))
B = np.zeros((2,2))

print("Horizontal concat:\n", np.hstack((A, B)))
print("Vertical concat:\n", np.vstack((A, B)))
```

```python
M = np.arange(1,17).reshape(4,4)
print("Matrix:\n", M)

print("First row:", M[0])
print("First column:", M[:,0])
print("2x2 block:\n", M[1:3, 1:3])
```
